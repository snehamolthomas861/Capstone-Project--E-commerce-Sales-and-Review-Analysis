# Handle missing values


*  Result of checking missing values using "df.isnull().sum()" shows that the dataset has missing values in 8 columns.

*  For categorical columns such as subcategory, material, size, and season, we can replace missing values with the mode.

*  For numerical columns such as rating, review_count, stock_quantity, and units_sold, we can use the median to fill missing values.





```python
df['subcategory'] = df['subcategory'].fillna(df['subcategory'].mode()[0])
```

```python
df['material'] = df['material'].fillna(df['material'].mode()[0])
```

```python
df['size'] = df['size'].fillna(df['size'].mode()[0])
```

```python
df['season'] = df['season'].fillna(df['season'].mode()[0])
```

```python
df['rating'] = df['rating'].fillna(df['rating'].median())
```

```python
df['review_count'] = df['review_count'].fillna(df['review_count'].median())
```

```python
df['stock_quantity'] = df['stock_quantity'].fillna(df['stock_quantity'].median())
```

```python
df['units_sold'] = df['units_sold'].fillna(df['units_sold'].median())
```

# Handle outliers

```python
outlier_columns = [
    'price_usd',
    'discount_percent',
    'review_count',
    'stock_quantity',
    'return_rate'
]

for column in outlier_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    Lower_limit = Q1 - 1.5 * IQR
    Upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < Lower_limit) |
        (df[column] > Upper_limit)
    ]

    print(f"\n{column}")
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Limit:", Lower_limit)
    print("Upper Limit:", Upper_limit)
    print("Number of Outliers:", len(outliers))
```

 However, several product prices exceeded this limit. These values were considered potential outliers, but they represent genuine product prices rather than incorrect data. Since higher-priced clothing items may naturally exist in the dataset, these values were retained without modification to preserve the accuracy and authenticity of the data. # **Data Visualization**


**Univariate analysis**

**1. Histogram: Product price distribution**



```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x='price_usd', bins=20, kde=True)

plt.title('Distribution of Clothing Product Prices')
plt.xlabel('Price (USD)')
plt.ylabel('Number of Products')

plt.show()
```

The dataset mainly consists of low- to moderately priced clothing products, while relatively expensive products are less common.

Most products are priced between approximately $ 20-$ 80, with the highest concentration around $30–$40. As the price increases, the number of products generally decreases.

Box Plot

**2. Distribution of Units Sold per Product**

```python
plt.figure(figsize=(8, 4))

sns.boxplot(data=df, x='units_sold')

plt.title('Distribution of Units Sold per Product')
plt.xlabel('Units Sold')

plt.tight_layout()
plt.show()
```

The box plot shows that the distribution of units sold per product is highly right-skewed. Most products have relatively low sales quantities, while a small number of products have exceptionally high sales with the highest value exceeding 10,000 units sold.

Count Plot

**3. Product Distribution across various Seasons**

```python
plt.figure(figsize=(8, 5))
order = df['season'].value_counts().index
sns.countplot(data=df, x='season', order=order)
plt.title('Product Distribution by Season')
plt.xlabel('Season')
plt.ylabel('Number of Products')
plt.tight_layout()
plt.show()
```

It shows the distribution of clothing products across different seasons.

Summer has the highest number of products, with values more than 1000, followed by Autumn and Winter.

Spring accounts for around less than 1000 products, while All Season has the lowest count.

**Bivariate analysis**

**4. Scatter plot: Product price vs units sold**

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='price_usd',
    y='units_sold',
    hue='category',
    alpha=0.7
)

plt.title('Product Price vs. Units Sold')
plt.xlabel('Price (USD)')
plt.ylabel('Units Sold')
plt.legend(title='Category', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

There is a strong concentration of high volume sales at lower price points.

The vast majority of high performing products(above 4,000 units) are under $50.

As product price increase, the volume of units sold drop significantly following a typical demand curve.

**5. Box plot: Units sold across product categories**

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x='category',
    y='units_sold'
)

plt.title('Units Sold Across Product Categories')
plt.xlabel('Product Category')
plt.ylabel('Units Sold')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

The vast majority of product categories have a very similar baseline distribution with volumes heavy clustered under 2,000 units.

**Dresses** show the highest sales spikes nearly 10,000 units.

**Jumpsuits and Coats** have lower sales volume when compared to other categories.

**6. Bar plot: Average return rate by material**

```python
material_returns = (
    df.groupby('material', as_index=False)['return_rate']
      .mean()
      .sort_values('return_rate', ascending=False)
)

plt.figure(figsize=(9, 5))

sns.barplot(
    data=material_returns,
    x='material',
    y='return_rate'
)

plt.title('Average Return Rate by Material')
plt.xlabel('Material')
plt.ylabel('Average Return Rate')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

**Nylon** is the best performing material with the lowest average return rate.

Heavy materials like Leather and wool also well keeping low return rates.

**Denim** has the highest average return rate among all other materials.

**7. Bar plot: Average rating by clothing style**

```python
style_ratings = (
    df.groupby('style', as_index=False)['rating']
      .mean()
      .sort_values('rating', ascending=False)
)

plt.figure(figsize=(9, 5))

sns.barplot(
    data=style_ratings,
    x='style',
    y='rating'
)

plt.title('Average Customer Rating by Clothing Style')
plt.xlabel('Style')
plt.ylabel('Average Rating')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

All clothing style shows an identical average customer rating around 3.0.

Streetwear has lower rating when compared to other styles and Formal has the highest rating comparitively.

**8. Scatter plot: Discount percentage vs. units sold**

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='discount_percent',
    y='units_sold',
    alpha=0.6
)

plt.title('Discount Percentage vs. Units Sold')
plt.xlabel('Discount (%)')
plt.ylabel('Units Sold')
plt.tight_layout()
plt.show()
```

Visually, there is no strong positive correlation between offering a higher discount and driving higher sales volume.

**10. Line Chart- Monthly Sales Trend**

```python
# Calculate total units sold per month
monthly_sales = df.groupby(
    df['OrderDate'].dt.to_period('M')
)['units_sold'].sum()

# Create line chart
fig = px.line(
    x=monthly_sales.index.to_timestamp(),
    y=monthly_sales.values,
    markers=True,
    title="Monthly Sales Trend",
    labels={
        "x": "Month and Year",
        "y": "Total Units Sold"
    }
)


fig.show()
```

It shows the monthly sales trend from January 2024 to July 2026. Sales fluctuate with several peaks and declines throughout the period.

The highest sales occur around between August-October 2025.However sales decrease significantly in early 2026.

It indicates that sales performance has become weaker in recent months.

**Multivariate analysis**

**12. Correlation heatmap: Numerical variables**

```python
numeric_cols = [
    'price_usd',
    'discount_percent',
    'rating',
    'review_count',
    'stock_quantity',
    'units_sold',
    'return_rate'
]

correlation_matrix = df[numeric_cols].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Numerical Variables')
plt.tight_layout()
plt.show()
```

Strong positive correlation between review_count and units_sold-> 0.91. This is the strongest relationship in the heatmap. Products with more reviews tend to have higher sales volumes.

Weak positive correlation between rating and review_count —> 0.21.
Products with more reviews show a slight tendency to have higher ratings.

Weak positive correlation between price_usd and return_rate —> 0.13. Higher-priced products may have slightly higher return rates, but the relationship is weak.

Most other variable pairs showed correlations close to zero, suggesting little or no linear relationship.

**13. Grouped bar chart: Seasonal sales by category**

```python
seasonal_sales = (
    df.groupby(['season', 'category'], as_index=False)['units_sold']
      .sum()
)

fig = px.bar(
    seasonal_sales,
    x='season',
    y='units_sold',
    color='category',
    barmode='group',
    title='Total Units Sold by Category and Season',
    labels={
        'season': 'Season',
        'units_sold': 'Total Units Sold',
        'category': 'Product Category'
    }
)

fig.show()
```

Summer has the highest sales for Activewear. This category performs exceptionally well in summer compared with other seasons.

Dresses perform well across multiple seasons, especially All Season,Summer and Winter

Sweaters show strong Winter sales,which is higher than their sales in other seasons.

Jackets perform well in Autumn and Winter.Some categories have lower sales in particular seasons, suggesting that demand varies by product type and season.
